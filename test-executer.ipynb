```python
import random
import os
import numpy as np
import json
import autotune_penalties as at
import problem_runner as pr

def set_global_seeds(seed):
    """Fija TODAS las semillas globales para reproducibilidad."""
    if seed is None:
        return
    random.seed(seed)
    np.random.seed(seed)
    os.environ["PYTHONHASHSEED"] = str(seed)

# 1) Cargar los mejores parámetros
# best_params_path = "autotune-results/q4rpd_full_fixed/best_params.json"
best_params_path = "autotune-results/medoids_improved/q4rpd_D21_P0_01/best_params.json"
# best_params_path = "autotune-results/individual_studies_first/q4rpd_D14_P1_00/best_params.json"
with open(best_params_path, encoding="utf-8") as f:
    best = json.load(f)["best_params"]

# 2) Construir problem_data: omega1=1.0, omega2 derivado, 6 penalizaciones
omega1 = 1.0
omega2 = omega1 * best["omega2_over_omega1"]
penalties = {k: best[k] for k in (
    "p_origin_penalty_w", "p_delivery_once_w", "p_position_once_w",
    "p_dest_inclusion_w", "p_consecutiveness_w", "p_capacity_w")}

INSTANCE = "D21_P0"   # cámbiala por la que quieras
seed=47
set_global_seeds(seed)
problem_data = at.build_problem_data(INSTANCE, omega1, omega2, penalties)

# 3) Ejecutar con SA COMPLETO (sin patched_sa_params) y build_srp_qubo TUNEADO
#    -> patched_qubo_builder aplica base_penalty_factor + los reward_*
with at.patched_qubo_builder(best):
    res = pr.run_problem(problem_data,
                         QCIUse=False, DWaveUse=False,
                         DWaveModel="", solver_type="",
                         dynamic_omega=False,
                         generate_graphics=True,   # dibuja las rutas
                         seed=seed, use_kmedoids=True, out_dir="test-executer")
print(res)
```
