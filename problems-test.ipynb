```python
import json, math
import autotune_penalties as at
import problem_runner as pr
import test_data

with open("autotune-results/q4rpd_full_fixed/best_params.json", encoding="utf-8") as f:
    best = json.load(f)["best_params"]

omega1 = 1.0
omega2 = omega1 * best["omega2_over_omega1"]
penalties = {k: best[k] for k in (
    "p_origin_penalty_w","p_delivery_once_w","p_position_once_w",
    "p_dest_inclusion_w","p_consecutiveness_w","p_capacity_w")}

INSTANCE = "D14_P1"
problem_data = at.build_problem_data(INSTANCE, omega1, omega2, penalties)

with at.patched_qubo_builder(best):
    res = pr.run_problem(problem_data, QCIUse=False, DWaveUse=False,
                         DWaveModel="", solver_type="", dynamic_omega=False,
                         generate_graphics=True, seed=0)

coords = test_data.VISUALIZATION_COORDS_DICT[INSTANCE]
leg = lambda a, b: math.hypot(coords[b][0]-coords[a][0], coords[b][1]-coords[a][1])
true_cost = sum(leg(r[i], r[i+1]) for r in res["routes"] for i in range(len(r)-1))
print(f"\nReportado : {res['total_route_duration']:.2f}")
print(f"Real (geo): {true_cost:.2f}")
print(f"Coinciden : {math.isclose(res['total_route_duration'], true_cost, rel_tol=1e-3)}")
print(f"Todas entregadas: {res['all_deliveries_done']}")
```
